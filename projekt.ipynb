# DICT

```python
# Pretvaranje DataFrame-a u rječnik
game_dict = {
    row['Title'].lower(): {  # Ključ je naziv igre u lowercase
        'Developer': row['Developer'],
        'Publisher': row['Publisher'],
        'Metascore': row['Metascore'],
        'Release Date': row['Release Date'],
        'Platforms': row['Platforms'].replace("Platforms\n", "").strip(),
        'Genre': row['Genre']
    }
    for i, row in df.iterrows()  # Iteriramo kroz svaki redak DataFrame-a
}

# Pretraživanje igre u rječniku
game = input('Upišite naziv igrice: ').strip().lower()

if game in game_dict:
    game_info = game_dict[game]
    print(f"Igrica {game.title()} pripada žanru {game_info['Genre']}. \n"
          f"Tvorac igrice je {game_info['Developer']}, a izdavač {game_info['Publisher']}. \n"
          f"Igrica ima ocjenu {game_info['Metascore']}, a dostupna je za igranje od {game_info['Release Date']}, "
          f"te se može igrati na {game_info['Platforms']}.")
else:
    print(f"Nema igrice sa nazivom {game.title()}.")
```

# FUNKCIJE

```python
# Funkcija za dohvaćanje detalja o igri

def get_game_details(game_name):
    game_row = df[df['Title'].str.lower() == game_name.strip().lower()]  # Case-insensitive pretraga
    if not game_row.empty:
        return {
            'Metascore': game_row['Metascore'].values[0],
            'Genre': game_row['Genre'].values[0],
            'Playtime': game_row['Playtime'].values[0]
        }
    else:
        return None

# Glavni dio programa
game = input("Unesite naziv igrice: ")
details = get_game_details(game)

if details:
    print(f"Igrica '{game}' ima Metascore {details['Metascore']}, \npripada žanru '{details['Genre']}', "
          f"\na prosječno vrijeme igranja je {details['Playtime']}.")
else:
    print(f"Nije pronađena igrica s nazivom '{game}'.")
```

# KORIŠTENJE GROUPBY

```python
# Grupiranje igara po izdavaču i izračun prosječnog Metascore-a (zaokruženo na dvije decimale)
grouped_by_publisher = df.groupby('Publisher')['Metascore'].mean().round(2)

print("Prosječan Metascore po izdavaču:")
print(grouped_by_publisher)
```

```python
# Grupiranje igara po žanru i brojanje koliko igara ima u svakom žanru
grouped_by_genre = df.groupby('Genre').size()

print("Broj igara po žanru:")
print(grouped_by_genre)
```

```python
# DGrupiranje po izdavaču i brojanje igara s ocjenom većom od 90
grouped_filtered = df[df['Metascore'] > 90].groupby('Publisher').size()

print("Broj igara po izdavaču s ocjenom većom od 90:")
print(grouped_filtered)
```

# GRAFOVI

```python
# Spider graf

# Grupiranje i selekcija top 8 žanrova prema Metascore
genre_grouped = df.groupby('Genre')['Metascore'].mean().dropna().sort_values(ascending=False)
top_genres = genre_grouped.head(8)
categories = list(top_genres.index)
values = list(top_genres.values)

# Dodajemo prvi element za zatvaranje grafa
categories.append(categories[0])
values.append(values[0])

# Postavljanje kutova za spider graf
angles = [i * 2 * 3.14159 / len(values) for i in range(len(values))]

# Postavljanje spider grafa
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Plotanje podataka
ax.plot(angles, values, linewidth=2, linestyle='solid', label='Metascore')
ax.fill(angles, values, alpha=0.4)

# Postavljanje oznaka
ax.set_xticks(angles)
ax.set_xticklabels(categories, fontsize=10)
ax.set_yticks(range(0, int(max(values)) + 1, 10))
ax.set_yticklabels(range(0, int(max(values)) + 1, 10), fontsize=8)

plt.title('Top 8 žanrova prema prosječnom Metascore-u', fontsize=16, pad=20)
plt.tight_layout()
plt.legend(loc='upper right')
plt.show()

```

```python
# Pivot tablica
heatmap_data = df.pivot_table(index='Publisher', columns='Age Rating', values='Metascore', aggfunc='mean')

# Filtriranje za top 10 izdavača prema broju igara
top_publishers = df['Publisher'].value_counts().head(10).index
heatmap_data = heatmap_data.loc[top_publishers]

# Kreiranje toplinskog grafa s drugačijim bojama
plt.figure(figsize=(12, 6))
plt.imshow(heatmap_data, cmap='viridis', aspect='auto')  # Viridis boje
plt.colorbar(label='Prosječan Metascore')

# Postavljanje oznaka
plt.xticks(ticks=range(len(heatmap_data.columns)), labels=heatmap_data.columns, rotation=45)
plt.yticks(ticks=range(len(heatmap_data.index)), labels=heatmap_data.index)
plt.title('Heatmap: Prosječan Metascore po izdavaču i dobnom rejtingu', fontsize=16)
plt.tight_layout()
plt.show()


```
